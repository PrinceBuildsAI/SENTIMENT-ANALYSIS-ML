## Convert text into number because ML mMdel olny understand numbers 

```python
df['emotion'].unique()
```

```python
unique_emotions = df['emotion'].unique()
emotion_numbers = {}
i = 0
for emo in unique_emotions:
    emotion_numbers[emo] = i
    i += 1

df['emotion'] = df['emotion'].map(emotion_numbers)
```

# text cleaning

```python
# convert lower case
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
#remove punchuation
import string
def remove_punc(txt):
    return txt.translate(str.maketrans('','',string.punctuation))

df['text'] = df['text'].apply(remove_punc)
```

```python
#removing numbers
def remove_numbers(txt):
    new =""
    for i in txt:
        if not i.isdigit():
            new = new+i
    return new

df['text'] = df['text'].apply(remove_numbers)
```

```python
#removing hyperlinks

import re

def remove_links(txt):
    return re.sub(r'http\S+|www\S+', '', txt)

df['text'] = df['text'].apply(remove_links)
```

```python
#removing html tags

def remove_html_tags(txt):
    return re.sub(r'<.*?>', '', txt)

df['text'] = df['text'].apply(remove_html_tags)
```

```python
#removing emojies     
#! but this has a limitation it only remove non ascii value it means it can olso removes (नमस्ते,  é → non-ASCII → removed)

def remove_emojis(txt):
    new=''
    for i in txt:
        if i.isascii():
            new += i 
    return new

df['text'] = df['text'].apply(remove_emojis)
```

```python
# Remove Stopwords - 'is', 'the', 'was'
#! using a library NLTK (Natural Language Toolkit)
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
len(stop_words)
```

```python
def remove(txt):
    words = txt.split()
    cleaned = []
    for i in words:
        if i not in stop_words:
            cleaned.append(i)
    return ' '.join(cleaned)

df['text'] = df['text'].apply(remove)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.20, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
```

# Uing Bag Of Words

```python
bow_vectorizer = CountVectorizer()
```

```python
#convert text into numerical representation
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
```

# using NB Algotihmh

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
nb_model = MultinomialNB()
```

```python
nb_model.fit(X_train_bow,y_train)
```

```python
pred_NB = nb_model.predict(X_test_bow)
print(accuracy_score(y_test, pred_NB))
```

# Using TF-IDF

```python
tfidf_vectorizer = TfidfVectorizer()
```

```python
#convert text into numerical representation
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)
```

```python
nb2_model = MultinomialNB()
nb2_model.fit(X_train_tfidf, y_train)
```

```python
pred_NB2 = nb2_model.predict(X_test_tfidf)
print(accuracy_score(y_test, pred_NB2))
```

## Using Logestic Regression to increase accuracy because Logestic Regression works on probabiliy

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_model = LogisticRegression(max_iter=1000)
```

```python
logistic_model.fit(X_train_tfidf, y_train)
```

```python
log_pred = logistic_model.predict(X_test_tfidf)
print(accuracy_score(y_test, log_pred))
```
